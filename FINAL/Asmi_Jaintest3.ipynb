```python
# parameter sentence
# 1. Create a function that takes a sentence as input, counts the number of vowels, and returns the count. 
def vowels(sen):
    # creating the vowels to iterate thorugh it 
    vowels ='aeiouAEIOU'
    # count arialbe is created 
    count = 0 
    for i in sen:
        if i in vowels:
            count = count + 1 
    return count 
sen = "this Is vowels"
# function call 
print(vowels(sen))

    


```

```python
''' 2. Implement a function that calculates the square root of each element in a list 
and returns a new list without inbuilt functions and use recursive function.'''



def sqroot(list1):
    new_list = []
    for i in list1:
        # to find the sqroot 
        num = i**0.5

        new_list.append(num)
    print(new_list)
    
list1 = [4,16,64]
sqroot(list1)
```

```python
''' 3. Create a generator function that yields the powers of 2 up to a certain limit.'''


def powe():
    i = 1 
    while True:
        # this is for power without inbuilt 
        yield i**2
        # we will increment it 
        i += 1 

obj = powe()
print(obj.__next__())
print(obj.__next__())
print(obj.__next__())

print(obj.__next__())
```

```python
'''5. Create a generator function that yields all the permutations of a given list.'''

# li = [1,2,3,4]
# for i in range(0,len(li)):
#     for j in range(i,len(li)):
#         print([li[i],li[j]])
# defiing function 
def perm(li):
    for i in range(0,len(li)):
        for j in range(i,len(li)):
            # yield to get value
            yield ([li[i],li[j]])
li = [1,2,3,4]
obj = perm(li)
print(obj.__next__())
print(obj.__next__())
print(obj.__next__())
print(obj.__next__())
print(obj.__next__())
print(obj.__next__())
print(obj.__next__())
print(obj.__next__())
print(obj.__next__())
print(obj.__next__())
```
