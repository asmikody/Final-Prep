```python
'''Que 4: Build a calculator that performs addition, subtraction, multiplication, or division
 based on user input. Handle division by zero and invalid operations.'''

from abc import ABC, abstractmethod

class Calculate(ABC):
    @abstractmethod
    def calculate(self, a, b, operation):
        pass

class Cal(Calculate):
    def __init__(self):
        self.history = []

    def add_to_history(self, opr, a, b, result):
        self.history.append(f"{a} {opr} {b} : {result}")

    def calculate(self, a, b, operation):
        if operation == '+':
            result = a + b
        elif operation == '-':
            result = a - b
        elif operation == '*':
            result = a * b
        elif operation == '/':
            result = a / b if b != 0 else "error: division by 0"
        else:
            return "Invalid Operation"

        self.add_to_history(operation, a, b, result)
        return result

    def get_history(self):
        return self.history
        

obj = Cal()
print(obj.calculate(2, 5, '-'))  # Outputs: -3
print(obj.calculate(10, 2, '/'))  # Outputs: 5.0
print(obj.get_history())  # Displays operation history
        
        
        
```

```python
'''Que 1: Write a program to manage a shopping cart using functions. Include a global cart_total variable and 
show how local and global variables work when adding items.'''

cart_total = 0
cart_items = []  

def add_to_cart(item, price): 
    global cart_total
    
    cart_total += price  
    
    cart_items.append((item, price))  
    
    print(f"Added '{item}' to the cart. Price: {price}")
    print(f"Updated cart total: {cart_total}\n") 
def remove_from_cart(item, price):
 
    global cart_total
    
    if (item, price) in cart_items:
        cart_items.remove((item, price))
        cart_total -= price  
        print(f"Removed '{item}' from the cart.")
        print(f"Updated cart total: {cart_total}\n")
    else:
        print(f"'{item}' not found in cart.\n")

def view_cart():
    print("Shopping Cart Items:")
    for item, price in cart_items:
        print(f"- {item}: {price}")
    print(f"Total Cost: {cart_total}\n")

# Test the program
add_to_cart("Laptop", 800)
add_to_cart("Mouse", 25)
remove_from_cart("Mouse", 25)
view_cart()
                 
```

```python
def calculate_average(filename):
    total_score = 0
    student_count = 0
    
    try:
        with open(filename, 'r') as f:
            for line in f:
                data = line.strip().split(',')

                # Skip completely invalid lines
                if len(data) < 2:
                    print(f"Skipping malformed line: {line.strip()}")
                    continue

                name, *scores = data  # Name followed by variable scores
                
                valid_scores = []
                                                                                                      
                # Validate each score
                for score in scores:
                    try:
                        valid_scores.append(float(score))
                    except ValueError:
                        print(f"Skipping invalid score for {name}: {score}")

                if valid_scores:
                    total_score += sum(valid_scores)
                    student_count += len(valid_scores)

        # Calculate and print average score
        if student_count > 0:
            avg_score = total_score / student_count
            print(f"Average Score: {avg_score:.2f}")
        else:
            print("No valid scores found.")

    except FileNotFoundError:
        print(f"Error: {filename} not found.")

# Example Usage
calculate_average("student.txt")
```
