```python
import re

def find_nearby_words(text, word1, word2, distance=5):
    # Convert text into a list of words
    words = text.split()
    
    # Loop through the words and check the distance
    for i in range(len(words)):
        if words[i] == word1:
            # Extract words within the given distance
            nearby_words = " ".join(words[max(0, i - distance): min(len(words), i + distance + 1)])
            
            # Search for the second word within the extracted range
            if re.search(rf"\b{word2}\b", nearby_words):
                print(f"'{word2}' found within {distance} words of '{word1}': {nearby_words}")

# Example usage
text = "The quick brown fox jumps over the lazy dog. The fox is very clever and quick."
find_nearby_words(text, "fox", "quick", 5)
```

```python
import re

# Sample URL
url = "https://example.com/page?name=John&age=30&city=NewYork"

# Regular expression with capturing groups and word boundaries
pattern = r"\bname=(?P<name>[^&]+)\b.*\bage=(?P<age>\d+)\b.*\bcity=(?P<city>[^&]+)\b"

# Using re.search() to extract values
match = re.search(pattern, url)

if match:
    name = match.group("name")
    age = match.group("age")
    city = match.group("city")

    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")
else:
    print("No match found")
```

```python
This pattern extracts name, age, and city using capturing groups and ensures that each parameter is treated as a distinct word.

Breaking it down:
\bname= → Ensures "name=" is matched as a whole word.

(?P<name>[^&]+) → Captures the value for name using a named group name. The [^&]+ ensures the match stops at &, preventing extra characters from being included.
1. Capturing Group ((?P<name>... ))
?P<name> is a named capturing group.

It assigns the extracted value to the name group.

Using a named group allows us to reference it later using match.group("name").

2. Character Class ([^&]+)
[^&] means "match any character except &".

+ ensures it matches one or more of those characters.

This prevents capturing extra parameters (&age=30), stopping at &.

Example Breakdown
python
import re

url = "https://example.com/page?name=John&age=30&city=NewYork"
pattern = r"name=(?P<name>[^&]+)"  # Capture name value

match = re.search(pattern, url)
if match:
    print(match.group("name"))  # Output: John
Summary
(?P<name>...) → Defines a named capturing group.

[^&]+ → Matches everything except &, ensuring the name value is correctly extracted.

Would you like a similar breakdown for other parts of your regex? 🚀


\bage= → Ensures "age=" is matched separately from other words.

(?P<age>\d+) → Captures digits only (\d+) for age.

\bcity= → Ensures "city=" is treated as a distinct word.

(?P<city>[^&]+) → Captures the city name, stopping at &.

4. Use re.search() to find matches
python
match = re.search(pattern, url)
re.search(pattern, url) looks for the first occurrence of the pattern in the URL.

It returns a Match object if successful, otherwise None.

5. Extract values using named capturing groups
python
if match:
    name = match.group("name")
    age = match.group("age")
    city = match.group("city")

    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")
else:
    print("No match found")
Explanation:
If a match is found:

match.group("name") → Retrieves the name (e.g., John).

match.group("age") → Retrieves the age (e.g., 30).

match.group("city") → Retrieves the city (e.g., NewYork).

These values are printed in a readable format.

6. Output
Name: John
Age: 30
City: NewYork
Would you like me to modify the pattern to dynamically handle other parameters as well? 🚀

```
